### Exception handling

keywords

try - use to wrap main logic

except - run after if try got some error 

else - run with try if no error


finally - runs no matter if we got error or not 

raise -


```python
num = 2
```

```python

try :
    print(10/num)
    print("division done")
except Exception as err:
    print("sorry, Something went wrong")
    print(f"Error : {err}")
else:
    print("good there's no Error")
finally:
    print("Process completed")
```

```python
age = 8
```

```python
try:
    if age < 10 or age > 18:
        raise ValueError("your age must be between 10 to 18")
    else:
        print("Welcome to the club")
except Exception as err:
    print(f"err : {err}")
```
