```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['GEMINI_API_KEY']=os.getenv('GEMINI_API_KEY')

from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI


load_dotenv()

def get_weather(city:str) -> str:
    """get the weater for a city."""
    return f"the weather in {city} is sunny."


model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)

agent = create_agent(
    model=model,
    tools=[get_weather],
    system_prompt="You are a very sarcastic agent."
)
```

```python
from pydantic import BaseModel, Field


class Color(BaseModel):
    name: str = Field(description="Name of the color")
    type: str = Field(description="Type of color, such as metallic, matte, or solid")


class Engine(BaseModel):
    type: str = Field(description="Type of engine")
    horsepower: int = Field(description="Engine horsepower")
    fuel_type: str = Field(description="Fuel type used by the engine")


class Car(BaseModel):
    brand: str = Field(description="Name of the car brand")
    model: str = Field(description="Model name of the car")
    top_speed: int = Field(description="Top speed of the car in km/h")

    engine: Engine = Field(
        description="Engine specifications of the car"
    )

    colors: list[Color] = Field(
        description="List of available colors for the car"
    )
```

```python
modal_with_structure = model.with_structured_output(Car)
```

```python
res = modal_with_structure.invoke("mustang")

res
```
