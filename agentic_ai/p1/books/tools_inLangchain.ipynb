```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['GEMINI_API_KEY']=os.getenv('GEMINI_API_KEY')

from langchain.agents import create_agent
from langchain_google_genai import ChatGoogleGenerativeAI


load_dotenv()

def get_weather(city:str) -> str:
    """get the weater for a city."""
    return f"the weather in {city} is sunny."


model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)

agent = create_agent(
    model=model,
    tools=[get_weather],
    system_prompt="You are a very sarcastic agent."
)
```

```python
from langchain.tools import tool

@tool
def get_weather(location):
    """Get the weather from provided location"""
    return f"it's 18degree c"


model_with_tools = model.bind_tools([get_weather])
```

```python
res = model_with_tools.invoke("what's the weather currently in delhi")

print(res)


# for tool_call in res.tool_calls:
#     print(f"tool : {tool_call['name']}")
    
```

```python
#testing

message = [
    {
    "role" : "system",
    "content" : "be professional,  if u'll get states explain it how its going to feel like if its weather tell that ull be feeling cold or hot or maybe something else "
},
    {
    "role" : "user",
    "content" : "what's the weather in delhi"
}]

ai_msg = model_with_tools.invoke(message)

message.append(ai_msg)

for tool_call in ai_msg.tool_calls:
    tool_result = get_weather.invoke(tool_call)
    message.append(tool_result)

print(message)

final = model_with_tools.invoke(message)

print(final.text)
```
